```python
# ==========================================================
# Section 1: Project Configuration & Load Frozen Datasets
# ==========================================================
#
# Purpose
# -------
# Initialise the project environment and load the frozen
# PolitikWeli and AfriHate datasets produced by:
#
#   • 01_politikweli_prepare.ipynb
#   • 02_afrihate_prepare.ipynb
#
# This notebook DOES NOT modify or re-split either dataset.
# It only integrates the corresponding frozen splits into
# the final multi-task dataset.
#
# Inputs
# ------
# data/processed/
#
# ├── politikweli/
# │     politikweli_master.csv
# │     politikweli_train.csv
# │     politikweli_validation.csv
# │     politikweli_test.csv
# │
# └── afrihate/
#       afrihate_master.csv
#       afrihate_train.csv
#       afrihate_validation.csv
#       afrihate_test.csv
#
# Outputs
# -------
# In-memory frozen datasets ready for integration.
#
# ==========================================================

from pathlib import Path
import logging
import random

import numpy as np
import pandas as pd

# ==========================================================
# Reproducibility
# ==========================================================

RANDOM_STATE = 42

random.seed(RANDOM_STATE)

np.random.seed(RANDOM_STATE)

# ==========================================================
# Project Structure
# ==========================================================

PROJECT_ROOT = Path.cwd().resolve()

DATA_DIR = PROJECT_ROOT / "data"

PROCESSED_DIR = DATA_DIR / "processed"

POLITIKWELI_DIR = PROCESSED_DIR / "politikweli"

AFRIHATE_DIR = PROCESSED_DIR / "afrihate"

MULTITASK_DIR = PROCESSED_DIR / "multitask"

REPORTS_DIR = PROJECT_ROOT / "reports"

for directory in [

    MULTITASK_DIR,

    REPORTS_DIR,

]:

    directory.mkdir(

        parents=True,

        exist_ok=True

    )

# ==========================================================
# Logging
# ==========================================================

logging.basicConfig(

    level=logging.INFO,

    format="%(asctime)s | %(levelname)s | %(message)s"

)

logger = logging.getLogger(__name__)

logger.info("=" * 70)
logger.info("MULTI-TASK DATASET CONSTRUCTION")
logger.info("=" * 70)

# ==========================================================
# Verify Frozen Input Datasets
# ==========================================================

required_files = {

    # ------------------------------------------------------
    # PolitikWeli
    # ------------------------------------------------------

    "PolitikWeli Master":

        POLITIKWELI_DIR /
        "politikweli_master.csv",

    "PolitikWeli Train":

        POLITIKWELI_DIR /
        "politikweli_train.csv",

    "PolitikWeli Validation":

        POLITIKWELI_DIR /
        "politikweli_validation.csv",

    "PolitikWeli Test":

        POLITIKWELI_DIR /
        "politikweli_test.csv",

    # ------------------------------------------------------
    # AfriHate
    # ------------------------------------------------------

    "AfriHate Master":

        AFRIHATE_DIR /
        "afrihate_master.csv",

    "AfriHate Train":

        AFRIHATE_DIR /
        "afrihate_train.csv",

    "AfriHate Validation":

        AFRIHATE_DIR /
        "afrihate_validation.csv",

    "AfriHate Test":

        AFRIHATE_DIR /
        "afrihate_test.csv",

}

missing_files = [

    name

    for name, path in required_files.items()

    if not path.exists()

]

if missing_files:

    raise FileNotFoundError(

        "The following required datasets were not found:\n\n"

        + "\n".join(missing_files)

        + "\n\n"

        "Please execute:\n"

        "  01_politikweli_prepare.ipynb\n"

        "  02_afrihate_prepare.ipynb\n"

        "before running this notebook."

    )

logger.info("✓ All frozen datasets located.")

# ==========================================================
# Load Frozen PolitikWeli Datasets
# ==========================================================

logger.info("=" * 70)
logger.info("LOADING POLITIKWELI DATASETS")
logger.info("=" * 70)

politikweli_master = pd.read_csv(

    required_files["PolitikWeli Master"],

    dtype={"text_id": str}

)

politikweli_train = pd.read_csv(

    required_files["PolitikWeli Train"],

    dtype={"text_id": str}

)

politikweli_validation = pd.read_csv(

    required_files["PolitikWeli Validation"],

    dtype={"text_id": str}

)

politikweli_test = pd.read_csv(

    required_files["PolitikWeli Test"],

    dtype={"text_id": str}

)

logger.info(

    f"Master:      {len(politikweli_master):,}"

)

logger.info(

    f"Train:       {len(politikweli_train):,}"

)

logger.info(

    f"Validation:  {len(politikweli_validation):,}"

)

logger.info(

    f"Test:        {len(politikweli_test):,}"

)

# ==========================================================
# Load Frozen AfriHate Datasets
# ==========================================================

logger.info("=" * 70)
logger.info("LOADING AFRIHATE DATASETS")
logger.info("=" * 70)

afrihate_master = pd.read_csv(

    required_files["AfriHate Master"]

)

afrihate_train = pd.read_csv(

    required_files["AfriHate Train"]

)

afrihate_validation = pd.read_csv(

    required_files["AfriHate Validation"]

)

afrihate_test = pd.read_csv(

    required_files["AfriHate Test"]

)

logger.info(

    f"Master:      {len(afrihate_master):,}"

)

logger.info(

    f"Train:       {len(afrihate_train):,}"

)

logger.info(

    f"Validation:  {len(afrihate_validation):,}"

)

logger.info(

    f"Test:        {len(afrihate_test):,}"

)

# ==========================================================
# Dataset Overview
# ==========================================================

logger.info("=" * 70)
logger.info("DATASET OVERVIEW")
logger.info("=" * 70)

overview = pd.DataFrame({

    "Dataset": [

        "PolitikWeli Master",

        "PolitikWeli Train",

        "PolitikWeli Validation",

        "PolitikWeli Test",

        "AfriHate Master",

        "AfriHate Train",

        "AfriHate Validation",

        "AfriHate Test",

    ],

    "Samples": [

        len(politikweli_master),

        len(politikweli_train),

        len(politikweli_validation),

        len(politikweli_test),

        len(afrihate_master),

        len(afrihate_train),

        len(afrihate_validation),

        len(afrihate_test),

    ]

})

print("\nFrozen Dataset Overview\n")

print(overview.to_string(index=False))

logger.info("=" * 70)
logger.info("SECTION 1 COMPLETE")
logger.info("=" * 70)

logger.info(

    "Frozen datasets loaded successfully."

)

logger.info(

    "Ready for compatibility and integrity validation."

)
```

```python
# ==========================================================
# Section 2: Dataset Compatibility & Integrity Audit
# ==========================================================
#
# Purpose
# -------
# Verify that the frozen PolitikWeli and AfriHate datasets
# are structurally compatible and suitable for integration.
#
# This section DOES NOT modify either dataset.
#
# It verifies:
#   • Required columns
#   • Missing values
#   • Duplicate texts
#   • Label integrity
#   • Split integrity
#   • Schema compatibility
#
# Outputs
# -------
# Console audit report.
#
# ==========================================================

logger.info("=" * 70)
logger.info("DATASET COMPATIBILITY & INTEGRITY AUDIT")
logger.info("=" * 70)

# ==========================================================
# Required Columns
# ==========================================================

POLITIKWELI_COLUMNS = {

    "text",
    "label",
    "label_text",
    "subset",
    "is_misinfo"

}

AFRIHATE_COLUMNS = {

    "clean_text",
    "label_id",
    "label_text",
    "is_strict_hate"

}

logger.info("Checking required columns...")

missing_pw = POLITIKWELI_COLUMNS.difference(
    politikweli_master.columns
)

missing_ah = AFRIHATE_COLUMNS.difference(
    afrihate_master.columns
)

if missing_pw:

    raise ValueError(

        "PolitikWeli is missing required columns:\n"

        f"{sorted(missing_pw)}"

    )

if missing_ah:

    raise ValueError(

        "AfriHate is missing required columns:\n"

        f"{sorted(missing_ah)}"

    )

logger.info("✓ Required columns verified.")

# ==========================================================
# Missing Values
# ==========================================================

logger.info("=" * 70)
logger.info("CHECKING MISSING VALUES")
logger.info("=" * 70)

print("\nPolitikWeli\n")

print(

    politikweli_master[
        ["text", "label", "label_text"]
    ].isnull().sum()

)

print("\nAfriHate\n")

print(

    afrihate_master[
        ["clean_text", "label_id", "label_text"]
    ].isnull().sum()

)

assert (

    politikweli_master["text"]

    .isna()

    .sum()

    == 0

)

assert (

    afrihate_master["clean_text"]

    .isna()

    .sum()

    == 0

)

logger.info("✓ No missing text detected.")

# ==========================================================
# Duplicate Text Audit
# ==========================================================

logger.info("=" * 70)
logger.info("CHECKING DUPLICATE TEXTS")
logger.info("=" * 70)

pw_duplicates = (

    politikweli_master["text"]

    .duplicated()

    .sum()

)

ah_duplicates = (

    afrihate_master["clean_text"]

    .duplicated()

    .sum()

)

print(

    f"\nPolitikWeli duplicate texts : "

    f"{pw_duplicates:,}"

)

print(

    f"AfriHate duplicate texts    : "

    f"{ah_duplicates:,}"

)

assert pw_duplicates == 0

assert ah_duplicates == 0

logger.info("✓ No duplicate texts detected.")

# ==========================================================
# Label Integrity
# ==========================================================

logger.info("=" * 70)
logger.info("VERIFYING LABEL INTEGRITY")
logger.info("=" * 70)

print("\nPolitikWeli Labels\n")

print(

    politikweli_master["label_text"]

    .value_counts()

    .sort_index()

)

print("\nAfriHate Labels\n")

print(

    afrihate_master["label_text"]

    .value_counts()

    .sort_index()

)

expected_pw = {

    "neutral",

    "fact",

    "fake"

}

expected_ah = {

    "normal",

    "abusive",

    "hate"

}

assert (

    set(

        politikweli_master["label_text"]

        .unique()

    )

    == expected_pw

)

assert (

    set(

        afrihate_master["label_text"]

        .unique()

    )

    == expected_ah

)

logger.info("✓ Label integrity verified.")

# ==========================================================
# Frozen Split Integrity
# ==========================================================

logger.info("=" * 70)
logger.info("VERIFYING FROZEN SPLITS")
logger.info("=" * 70)

def verify_split_overlap(

    train_df,

    validation_df,

    test_df,

    text_column,

    dataset_name,

):

    train = set(train_df[text_column])

    validation = set(validation_df[text_column])

    test = set(test_df[text_column])

    assert len(train & validation) == 0

    assert len(train & test) == 0

    assert len(validation & test) == 0

    logger.info(

        f"✓ {dataset_name}: "

        "No overlap detected."

    )

verify_split_overlap(

    politikweli_train,

    politikweli_validation,

    politikweli_test,

    "text",

    "PolitikWeli",

)

verify_split_overlap(

    afrihate_train,

    afrihate_validation,

    afrihate_test,

    "clean_text",

    "AfriHate",

)

# ==========================================================
# Schema Compatibility
# ==========================================================

logger.info("=" * 70)
logger.info("VERIFYING SCHEMA COMPATIBILITY")
logger.info("=" * 70)

comparison = pd.DataFrame({

    "PolitikWeli": sorted(

        politikweli_master.columns

    ),

    "AfriHate": sorted(

        afrihate_master.columns

    )

})

print("\nColumn Comparison\n")

print(comparison)

logger.info(

    "Schemas successfully inspected."

)

print(

    "\n✓ PolitikWeli passed compatibility checks."

)

print(

    "✓ AfriHate passed compatibility checks."

)

print(

    "✓ Frozen datasets ready for integration."

)

logger.info("=" * 70)
logger.info("SECTION 2 COMPLETE")
logger.info("=" * 70)
```

```python
# ==========================================================
# Section 3: Construct Multi-task Targets
# ==========================================================
#
# Purpose
# -------
# Transform the frozen PolitikWeli and AfriHate datasets into
# a common multi-task learning format.
#
# This section:
#   • Creates working copies of each dataset
#   • Harmonises column names
#   • Creates task identifiers
#   • Creates task-specific labels
#   • Adds binary targets
#   • Standardises metadata
#
# NOTE
# ----
# The original frozen datasets are never modified.
#
# Outputs
# -------
# Eight harmonised dataframes:
#
#   politikweli_mt_*
#   afrihate_mt_*
#
# ready for integration.
#
# ==========================================================

logger.info("=" * 70)
logger.info("CONSTRUCTING MULTI-TASK TARGETS")
logger.info("=" * 70)

# ==========================================================
# Create Working Copies
# ==========================================================

logger.info("Creating working copies...")

politikweli_mt_master = politikweli_master.copy()
politikweli_mt_train = politikweli_train.copy()
politikweli_mt_validation = politikweli_validation.copy()
politikweli_mt_test = politikweli_test.copy()

afrihate_mt_master = afrihate_master.copy()
afrihate_mt_train = afrihate_train.copy()
afrihate_mt_validation = afrihate_validation.copy()
afrihate_mt_test = afrihate_test.copy()

logger.info("✓ Working copies created.")

# ==========================================================
# Helper Function
# ==========================================================

def prepare_multitask_dataset(df, dataset_name):

    """
    Convert a prepared dataset into the common
    multi-task schema.
    """

    df = df.copy()

    # ------------------------------------------------------
    # Standardise text column
    # ------------------------------------------------------

    if dataset_name == "politikweli":

        df = df.rename(

            columns={

                "text": "text"

            }

        )

        df["task"] = "misinformation"

        df["dataset"] = "PolitikWeli"

        df["task_label"] = df["label"]

        df["task_label_text"] = df["label_text"]

        df["binary_target"] = df["is_misinfo"]

    elif dataset_name == "afrihate":

        df = df.rename(

            columns={

                "clean_text": "text",

                "label_id": "task_label",

                "label_text": "task_label_text"

            }

        )

        df["task"] = "hate_speech"

        df["dataset"] = "AfriHate"

        df["binary_target"] = df["is_strict_hate"]

    else:

        raise ValueError("Unknown dataset.")

    return df

# ==========================================================
# Harmonise PolitikWeli
# ==========================================================

logger.info("=" * 70)
logger.info("PROCESSING POLITIKWELI")
logger.info("=" * 70)

politikweli_mt_master = prepare_multitask_dataset(

    politikweli_mt_master,

    "politikweli"

)

politikweli_mt_train = prepare_multitask_dataset(

    politikweli_mt_train,

    "politikweli"

)

politikweli_mt_validation = prepare_multitask_dataset(

    politikweli_mt_validation,

    "politikweli"

)

politikweli_mt_test = prepare_multitask_dataset(

    politikweli_mt_test,

    "politikweli"

)

logger.info("✓ PolitikWeli harmonised.")

# ==========================================================
# Harmonise AfriHate
# ==========================================================

logger.info("=" * 70)
logger.info("PROCESSING AFRIHATE")
logger.info("=" * 70)

afrihate_mt_master = prepare_multitask_dataset(

    afrihate_mt_master,

    "afrihate"

)

afrihate_mt_train = prepare_multitask_dataset(

    afrihate_mt_train,

    "afrihate"

)

afrihate_mt_validation = prepare_multitask_dataset(

    afrihate_mt_validation,

    "afrihate"

)

afrihate_mt_test = prepare_multitask_dataset(

    afrihate_mt_test,

    "afrihate"

)

logger.info("✓ AfriHate harmonised.")

# ==========================================================
# Verify Common Schema
# ==========================================================

logger.info("=" * 70)
logger.info("VERIFYING COMMON SCHEMA")
logger.info("=" * 70)

required_columns = [

    "text",

    "task",

    "dataset",

    "task_label",

    "task_label_text",

    "binary_target"

]

for column in required_columns:

    assert column in politikweli_mt_master.columns

    assert column in afrihate_mt_master.columns

logger.info("✓ Common schema verified.")

# ==========================================================
# Dataset Summary
# ==========================================================

summary = pd.DataFrame({

    "Dataset": [

        "PolitikWeli",

        "AfriHate"

    ],

    "Samples": [

        len(politikweli_mt_master),

        len(afrihate_mt_master)

    ],

    "Task": [

        "Misinformation",

        "Hate Speech"

    ],

    "Classes": [

        politikweli_mt_master["task_label_text"].nunique(),

        afrihate_mt_master["task_label_text"].nunique()

    ]

})

print("\nMulti-task Dataset Summary\n")

print(summary.to_string(index=False))

logger.info("=" * 70)
logger.info("SECTION 3 COMPLETE")
logger.info("=" * 70)

logger.info(

    "Both datasets successfully converted to the common "

    "multi-task schema."

)
```

```python
# ==========================================================
# Section 4: Merge Frozen Splits & Export Multi-task Dataset
# ==========================================================
#
# Purpose
# -------
# Merge the frozen PolitikWeli and AfriHate splits into the
# final multi-task datasets used for model development.
#
# This section:
#   • Merges corresponding frozen splits
#   • Verifies split integrity
#   • Computes task-specific class weights
#   • Exports final datasets
#   • Saves metadata and audit reports
#
# Outputs
# -------
# data/processed/multitask/
#
#     multitask_master.csv
#     multitask_train.csv
#     multitask_validation.csv
#     multitask_test.csv
#     multitask_dataset_summary.csv
#     multitask_class_weights.json
#     multitask_metadata.json
#     multitask_audit.json
#
# ==========================================================

import json
from sklearn.utils.class_weight import compute_class_weight

logger.info("=" * 70)
logger.info("MERGING FROZEN MULTI-TASK DATASETS")
logger.info("=" * 70)

# ==========================================================
# Merge Frozen Splits
# ==========================================================

multitask_master = pd.concat(

    [

        politikweli_mt_master,

        afrihate_mt_master

    ],

    ignore_index=True

)

multitask_train = pd.concat(

    [

        politikweli_mt_train,

        afrihate_mt_train

    ],

    ignore_index=True

)

multitask_validation = pd.concat(

    [

        politikweli_mt_validation,

        afrihate_mt_validation

    ],

    ignore_index=True

)

multitask_test = pd.concat(

    [

        politikweli_mt_test,

        afrihate_mt_test

    ],

    ignore_index=True

)

logger.info("✓ Frozen splits merged.")

# ==========================================================
# Shuffle Each Split
# ==========================================================

multitask_master = multitask_master.sample(

    frac=1,

    random_state=RANDOM_STATE

).reset_index(drop=True)

multitask_train = multitask_train.sample(

    frac=1,

    random_state=RANDOM_STATE

).reset_index(drop=True)

multitask_validation = multitask_validation.sample(

    frac=1,

    random_state=RANDOM_STATE

).reset_index(drop=True)

multitask_test = multitask_test.sample(

    frac=1,

    random_state=RANDOM_STATE

).reset_index(drop=True)

logger.info("✓ Splits shuffled.")

# ==========================================================
# Verify Split Integrity
# ==========================================================

logger.info("=" * 70)
logger.info("VERIFYING FINAL SPLITS")
logger.info("=" * 70)

train_texts = set(multitask_train["text"])

validation_overlap = len(

    train_texts.intersection(

        multitask_validation["text"]

    )

)

test_overlap = len(

    train_texts.intersection(

        multitask_test["text"]

    )

)

assert validation_overlap == 0

assert test_overlap == 0

logger.info("✓ No train-validation leakage.")

logger.info("✓ No train-test leakage.")

# ==========================================================
# Dataset Summary
# ==========================================================

summary = pd.DataFrame({

    "Split": [

        "Master",

        "Train",

        "Validation",

        "Test"

    ],

    "Samples": [

        len(multitask_master),

        len(multitask_train),

        len(multitask_validation),

        len(multitask_test)

    ]

})

print("\nFinal Multi-task Dataset Summary\n")

print(summary)

# ==========================================================
# Compute Task-specific Class Weights
# ==========================================================

logger.info("Computing training class weights...")

class_weights = {}

# ----------------------------------------------------------
# PolitikWeli
# ----------------------------------------------------------

pw_train = multitask_train[

    multitask_train["dataset"] == "PolitikWeli"

]

weights = compute_class_weight(

    class_weight="balanced",

    classes=np.sort(

        pw_train["task_label"].unique()

    ),

    y=pw_train["task_label"]

)

class_weights["PolitikWeli"] = {

    str(cls): float(weight)

    for cls, weight in zip(

        np.sort(

            pw_train["task_label"].unique()

        ),

        weights

    )

}

# ----------------------------------------------------------
# AfriHate
# ----------------------------------------------------------

ah_train = multitask_train[

    multitask_train["dataset"] == "AfriHate"

]

weights = compute_class_weight(

    class_weight="balanced",

    classes=np.sort(

        ah_train["task_label"].unique()

    ),

    y=ah_train["task_label"]

)

class_weights["AfriHate"] = {

    str(cls): float(weight)

    for cls, weight in zip(

        np.sort(

            ah_train["task_label"].unique()

        ),

        weights

    )

}

logger.info("✓ Class weights computed.")

# ==========================================================
# Export Final Datasets
# ==========================================================

logger.info("=" * 70)
logger.info("EXPORTING DATASETS")
logger.info("=" * 70)

multitask_master.to_csv(

    MULTITASK_DIR /

    "multitask_master.csv",

    index=False

)

multitask_train.to_csv(

    MULTITASK_DIR /

    "multitask_train.csv",

    index=False

)

multitask_validation.to_csv(

    MULTITASK_DIR /

    "multitask_validation.csv",

    index=False

)

multitask_test.to_csv(

    MULTITASK_DIR /

    "multitask_test.csv",

    index=False

)

summary.to_csv(

    MULTITASK_DIR /

    "multitask_dataset_summary.csv",

    index=False

)

with open(

    MULTITASK_DIR /

    "multitask_class_weights.json",

    "w",

    encoding="utf-8"

) as fp:

    json.dump(

        class_weights,

        fp,

        indent=4

    )

# ==========================================================
# Metadata
# ==========================================================

metadata = {

    "dataset": "PolitikWeli + AfriHate Multi-task",

    "random_state": RANDOM_STATE,

    "construction": "Frozen split merge",

    "source_notebooks": [

        "01_politikweli_prepare.ipynb",

        "02_afrihate_prepare.ipynb"

    ],

    "splits": {

        "master": len(multitask_master),

        "train": len(multitask_train),

        "validation": len(multitask_validation),

        "test": len(multitask_test)

    }

}

with open(

    MULTITASK_DIR /

    "multitask_metadata.json",

    "w",

    encoding="utf-8"

) as fp:

    json.dump(

        metadata,

        fp,

        indent=4

    )

# ==========================================================
# Audit Report
# ==========================================================

audit = {

    "duplicate_texts": int(

        multitask_master["text"]

        .duplicated()

        .sum()

    ),

    "missing_text": int(

        multitask_master["text"]

        .isna()

        .sum()

    ),

    "datasets": multitask_master["dataset"]

        .value_counts()

        .to_dict(),

    "tasks": multitask_master["task"]

        .value_counts()

        .to_dict()

}

with open(

    MULTITASK_DIR /

    "multitask_audit.json",

    "w",

    encoding="utf-8"

) as fp:

    json.dump(

        audit,

        fp,

        indent=4

    )

# ==========================================================
# Final Validation
# ==========================================================

logger.info("=" * 70)
logger.info("FINAL VALIDATION")
logger.info("=" * 70)

generated_files = [

    "multitask_master.csv",

    "multitask_train.csv",

    "multitask_validation.csv",

    "multitask_test.csv",

    "multitask_dataset_summary.csv",

    "multitask_class_weights.json",

    "multitask_metadata.json",

    "multitask_audit.json"

]

for file in generated_files:

    path = MULTITASK_DIR / file

    if path.exists():

        logger.info(f"✓ {file}")

    else:

        logger.warning(f"✗ {file}")

logger.info("=" * 70)
logger.info("MULTI-TASK DATASET CONSTRUCTION COMPLETE")
logger.info("=" * 70)

print("\nGenerated Files\n")

for file in sorted(MULTITASK_DIR.glob("*")):

    print(f"✓ {file.name}")

print("\nFinal Dataset Sizes")

print(f"Master      : {len(multitask_master):,}")

print(f"Train       : {len(multitask_train):,}")

print(f"Validation  : {len(multitask_validation):,}")

print(f"Test        : {len(multitask_test):,}")

logger.info("Notebook 03 completed successfully.")
```

```python
# ==========================================================
# Section 5: Final Multi-task Audit & Export
# ==========================================================
#
# Purpose
# -------
# Perform final quality assurance on the multi-task datasets,
# compute task-specific class weights, generate audit reports,
# and export all datasets and reproducibility artefacts.
#
# Outputs
# -------
# data/processed/multitask/
#
#     multitask_master.csv
#     multitask_train.csv
#     multitask_validation.csv
#     multitask_test.csv
#     multitask_dataset_summary.csv
#     multitask_class_weights.json
#     multitask_metadata.json
#     multitask_audit.json
#
# ==========================================================

import json
from sklearn.utils.class_weight import compute_class_weight

logger.info("=" * 70)
logger.info("FINAL MULTI-TASK AUDIT & EXPORT")
logger.info("=" * 70)

# ==========================================================
# Dataset Summary
# ==========================================================

logger.info("Generating dataset summary...")

summary = pd.DataFrame({

    "Split": [

        "Master",

        "Train",

        "Validation",

        "Test"

    ],

    "Samples": [

        len(multitask_master),

        len(multitask_train),

        len(multitask_validation),

        len(multitask_test)

    ]

})

print("\nOverall Split Summary\n")

print(summary)

# ==========================================================
# Dataset Contribution by Source
# ==========================================================

logger.info("Verifying dataset contribution...")

split_distribution = pd.DataFrame({

    "Train":

        multitask_train["dataset"]

        .value_counts(),

    "Validation":

        multitask_validation["dataset"]

        .value_counts(),

    "Test":

        multitask_test["dataset"]

        .value_counts()

}).fillna(0).astype(int)

print("\nDataset Contribution\n")

print(split_distribution)

# ==========================================================
# Task Distribution
# ==========================================================

task_distribution = pd.DataFrame({

    "Master":

        multitask_master["task"]

        .value_counts(),

    "Train":

        multitask_train["task"]

        .value_counts(),

    "Validation":

        multitask_validation["task"]

        .value_counts(),

    "Test":

        multitask_test["task"]

        .value_counts()

}).fillna(0).astype(int)

print("\nTask Distribution\n")

print(task_distribution)

# ==========================================================
# Compute Training Class Weights
# ==========================================================

logger.info("Computing task-specific class weights...")

class_weights = {}

# ----------------------------------------------------------
# PolitikWeli
# ----------------------------------------------------------

pw_train = multitask_train[

    multitask_train["dataset"] == "PolitikWeli"

]

pw_weights = compute_class_weight(

    class_weight="balanced",

    classes=np.sort(

        pw_train["task_label"].unique()

    ),

    y=pw_train["task_label"]

)

class_weights["PolitikWeli"] = {

    str(label): float(weight)

    for label, weight in zip(

        np.sort(

            pw_train["task_label"].unique()

        ),

        pw_weights

    )

}

# ----------------------------------------------------------
# AfriHate
# ----------------------------------------------------------

ah_train = multitask_train[

    multitask_train["dataset"] == "AfriHate"

]

ah_weights = compute_class_weight(

    class_weight="balanced",

    classes=np.sort(

        ah_train["task_label"].unique()

    ),

    y=ah_train["task_label"]

)

class_weights["AfriHate"] = {

    str(label): float(weight)

    for label, weight in zip(

        np.sort(

            ah_train["task_label"].unique()

        ),

        ah_weights

    )

}

logger.info("✓ Training class weights computed.")

# ==========================================================
# Export Datasets
# ==========================================================

logger.info("Exporting datasets...")

multitask_master.to_csv(

    MULTITASK_DIR /

    "multitask_master.csv",

    index=False

)

multitask_train.to_csv(

    MULTITASK_DIR /

    "multitask_train.csv",

    index=False

)

multitask_validation.to_csv(

    MULTITASK_DIR /

    "multitask_validation.csv",

    index=False

)

multitask_test.to_csv(

    MULTITASK_DIR /

    "multitask_test.csv",

    index=False

)

summary.to_csv(

    MULTITASK_DIR /

    "multitask_dataset_summary.csv",

    index=False

)

logger.info("✓ Datasets exported.")

# ==========================================================
# Save Class Weights
# ==========================================================

with open(

    MULTITASK_DIR /

    "multitask_class_weights.json",

    "w",

    encoding="utf-8"

) as fp:

    json.dump(

        class_weights,

        fp,

        indent=4

    )

# ==========================================================
# Metadata
# ==========================================================

metadata = {

    "dataset": "PolitikWeli + AfriHate Multi-task",

    "construction_method":

        "Merge frozen train/validation/test splits",

    "random_state": RANDOM_STATE,

    "source_notebooks": [

        "01_politikweli_prepare.ipynb",

        "02_afrihate_prepare.ipynb",

        "03_build_multitask_dataset.ipynb"

    ],

    "datasets": {

        "PolitikWeli":

            len(politikweli_master),

        "AfriHate":

            len(afrihate_master)

    },

    "splits": {

        "train":

            len(multitask_train),

        "validation":

            len(multitask_validation),

        "test":

            len(multitask_test)

    }

}

with open(

    MULTITASK_DIR /

    "multitask_metadata.json",

    "w",

    encoding="utf-8"

) as fp:

    json.dump(

        metadata,

        fp,

        indent=4

    )

# ==========================================================
# Audit Report
# ==========================================================

audit = {

    "master_records":

        int(len(multitask_master)),

    "duplicate_texts":

        int(

            multitask_master["text"]

            .duplicated()

            .sum()

        ),

    "missing_text":

        int(

            multitask_master["text"]

            .isna()

            .sum()

        ),

    "missing_task_labels":

        int(

            multitask_master["task_label"]

            .isna()

            .sum()

        ),

    "dataset_distribution":

        multitask_master["dataset"]

        .value_counts()

        .to_dict(),

    "task_distribution":

        multitask_master["task"]

        .value_counts()

        .to_dict()

}

with open(

    MULTITASK_DIR /

    "multitask_audit.json",

    "w",

    encoding="utf-8"

) as fp:

    json.dump(

        audit,

        fp,

        indent=4

    )

# ==========================================================
# Final Validation
# ==========================================================

logger.info("=" * 70)
logger.info("FINAL VALIDATION")
logger.info("=" * 70)

expected_outputs = [

    "multitask_master.csv",

    "multitask_train.csv",

    "multitask_validation.csv",

    "multitask_test.csv",

    "multitask_dataset_summary.csv",

    "multitask_class_weights.json",

    "multitask_metadata.json",

    "multitask_audit.json"

]

for filename in expected_outputs:

    path = MULTITASK_DIR / filename

    if path.exists():

        logger.info(f"✓ {filename}")

    else:

        logger.warning(f"✗ {filename}")

logger.info("=" * 70)
logger.info("MULTI-TASK DATASET CONSTRUCTION COMPLETE")
logger.info("=" * 70)

print("\nGenerated Files\n")

for file in sorted(MULTITASK_DIR.glob("*")):

    print(f"✓ {file.name}")

print("\nFinal Dataset Sizes\n")

print(f"Master      : {len(multitask_master):,}")

print(f"Train       : {len(multitask_train):,}")

print(f"Validation  : {len(multitask_validation):,}")

print(f"Test        : {len(multitask_test):,}")

logger.info(

    "Notebook 03 completed successfully."

)

logger.info(

    "Final multi-task dataset ready for model training."

)
```
